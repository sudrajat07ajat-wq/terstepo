### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
# Import library database dan pandas
import sqlite3
import pandas as pd

# 1. Hubungkan langsung ke file database 'my_data1.db'
con = sqlite3.connect("my_data1.db")

# 2. Tulis query SQL-nya
query = 'SELECT DISTINCT "LAUNCH_SITE" FROM SPACEXTABLE'

# 3. Minta Pandas untuk menjalankan query dan menampilkan hasilnya
df = pd.read_sql(query, con)
df
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'KSC' 


```python
query = "SELECT * FROM SPACEXTABLE WHERE LAUNCH_SITE LIKE 'KSC%' LIMIT 5"
pd.read_sql(query, con)
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
query = "SELECT SUM(PAYLOAD_MASS__KG_) as Total_Payload FROM SPACEXTABLE WHERE CUSTOMER = 'NASA (CRS)'"
pd.read_sql(query, con)
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
query = "SELECT AVG(PAYLOAD_MASS__KG_) as Average_Payload FROM SPACEXTABLE WHERE BOOSTER_VERSION = 'F9 v1.1'"
pd.read_sql(query, con)
```

### Task 5

##### List the date where the succesful landing outcome in drone ship was acheived.


_Hint:Use min function_ 


```python
query = "SELECT MIN(DATE) as First_Success_Date FROM SPACEXTABLE WHERE LANDING_OUTCOME = 'Success (drone ship)'"
pd.read_sql(query, con)
```

### Task 6

##### List the names of the boosters which have success in ground pad  and have payload mass greater than 4000 but less than 6000


```python
query = """
    SELECT BOOSTER_VERSION
    FROM SPACEXTABLE
    WHERE LANDING_OUTCOME = 'Success (ground pad)'
    AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000
"""
pd.read_sql(query, con)
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
query = "SELECT LANDING_OUTCOME, COUNT(LANDING_OUTCOME) as Count FROM SPACEXTABLE GROUP BY LANDING_OUTCOME"
pd.read_sql(query, con)
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass. Use a subquery.


```python
query = """
    SELECT BOOSTER_VERSION
    FROM SPACEXTABLE
    WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE)
"""
pd.read_sql(query, con)
```

### Task 9


##### List the records which will display the month names, succesful landing_outcomes in ground pad ,booster versions, launch_site for the months in year 2017
**Note: SQLLite does not support monthnames. So you need to use substr(Date,6,2) for month, substr(Date,9,2) for date, substr(Date,0,5),='2017' for year.**


```python
query = """
    SELECT DATE, LANDING_OUTCOME, BOOSTER_VERSION, LAUNCH_SITE
    FROM SPACEXTABLE
    WHERE LANDING_OUTCOME = 'Success (ground pad)'
    AND substr(Date,0,5) = '2017'
"""
df_task9 = pd.read_sql(query, con)

if not df_task9.empty:
    df_task9['Date_Object'] = pd.to_datetime(df_task9['Date'])
    df_task9['Month_Name'] = df_task9['Date_Object'].dt.month_name()

    # 3. Tampilkan (Gunakan Nama Kolom yang Benar: Huruf Kecil/Title Case)
    # Perhatikan: 'Landing_Outcome', 'Booster_Version', 'Launch_Site'
    print(df_task9[['Month_Name', 'Landing_Outcome', 'Booster_Version', 'Launch_Site']])
else:
    print("Data tidak ditemukan.")
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order


```python
# Task 10 (Perbaikan Nama Kolom)

# 1. Ambil data (Nama kolom di Query SQL boleh huruf besar, tapi di Pandas nanti jadi Title Case)
query = "SELECT LANDING_OUTCOME, DATE FROM SPACEXTABLE"
df_task10 = pd.read_sql(query, con)

df_task10['Date'] = pd.to_datetime(df_task10['Date'])

start_date = pd.to_datetime('2010-06-04')
end_date = pd.to_datetime('2017-03-20')

filtered_data = df_task10[
    (df_task10['Date'] >= start_date) &
    (df_task10['Date'] <= end_date)
]

ranking = filtered_data['Landing_Outcome'].value_counts()
print(ranking)
```
